```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

pd.options.display.float_format = '{:.2f}'.format

HS_data = pd.read_csv('AICPA_regressionAnalysisData.csv')

display(HS_data)
```

```python
HS_data['date'] = pd.to_datetime(HS_data['date'])
display(HS_data)
```

Winter is December, Jan, Feb
Month 12, 1, 2

```python
HS_data['winter_DV'] = np.where(HS_data['date'].dt.month.isin([12,1,2]),1,0)

display(HS_data)
```

```python
# interaction i the independant variable * the dummy variable
HS_data['winter_interaction'] = HS_data['production'] * HS_data['winter_DV']
display(HS_data)
```

```python
HS_data['fall_DV'] = np.where(HS_data['date'].dt.month.isin([9,10,11]),1,0)

display(HS_data)
```

```python
HS_data['fall_interaction'] = HS_data['production'] * HS_data['fall_DV']
display(HS_data)
```

```python
dt4training = HS_data.loc[HS_data['type']=='dt4training']
dt4testing = HS_data.loc[HS_data['type']=='dt4testing']

display(dt4training)
display(dt4testing)
```

```python
y = dt4training.loc[:, 'revenue']
x = dt4training.loc[:, ['production', 'winter_DV','winter_interaction']]

x = sm.add_constant(x)

model1 = sm.OLS(y, x).fit()

model1.params
```

```python
y = dt4training.loc[:, 'revenue']
x = dt4training.loc[:, ['production', 'fall_DV','fall_interaction']]

x = sm.add_constant(x)

model2 = sm.OLS(y, x).fit()

model2.params
```

```python
x1_test = sm.add_constant(dt4testing.loc[:, ['production', 'winter_DV', 'winter_interaction']])
x2_test = sm.add_constant(dt4testing.loc[:, ['production', 'fall_DV', 'fall_interaction']])

model_assessment = dt4testing.copy()
model_assessment['model1_predicted_revenue'] = model1.predict(x1_test)
model_assessment['model2_predicted_revenue'] = model2.predict(x2_test)

display(model_assessment)
```

```python
plt.figure(figsize=(10,6))

plt.title('revenue forecase with winter DV')
plt.ylabel('revenue')
plt.xlabel('production')
# visualize the regrssion w/ DV
plt.scatter(dt4training['production'], dt4training['revenue'])
# extract coefficents
intercept, production_coefficient, winter_DV_intercept, winter_coefficient = model1.params
# create lines for non winter and winter
non_winter_line = intercept + production_coefficient * dt4training['production']
winter_line = (intercept + winter_DV_intercept) + (production_coefficient + winter_coefficient) *dt4training['production']
# plot y values agianst x values for the inercept
plt.plot(dt4training['production'], non_winter_line,color='red')
plt.plot(dt4training['production'], winter_line,color='blue')
```

```python
plt.figure(figsize=(10,6))

plt.title('revenue forecase with fall DV')
plt.ylabel('revenue')
plt.xlabel('production')
# visualize the regrssion w/ DV
plt.scatter(dt4training['production'], dt4training['revenue'])
# extract coefficents
intercept, production_coefficient, fall_DV_intercept, fall_coefficient = model1.params
# create lines for non winter and winter
non_fall_line = intercept + production_coefficient * dt4training['production']
fall_line = (intercept + fall_DV_intercept) + (production_coefficient + fall_coefficient) *dt4training['production']
# plot y values agianst x values for the inercept
plt.plot(dt4training['production'], non_fall_line,color='red')
plt.plot(dt4training['production'], fall_line,color='blue')
```

```python
# for Production
y = dt4training.loc[:, 'revenue']
x = dt4training.loc[:, ['production']]

x = sm.add_constant(x)

model1 = sm.OLS(y, x).fit()
model1.params

x_test = dt4testing.loc[:, ['production']]
x_test = sm.add_constant(x_test)

model1_assessment = dt4testing.copy()

model1_assessment['model1_predicted_revenue'] = model1.predict(x_test)
model1_assessment['model1_abs_pct_error'] = abs(
    (model1_assessment['revenue'] - model1_assessment['model1_predicted_revenue']
    ) / model1_assessment['revenue'])

display(model1_assessment)

model1_assessment['model1_abs_pct_error'].mean()
```

```python
model_assessment['model1_abs_pct_error'] = abs(
    (model_assessment['revenue'] - model_assessment['model1_predicted_revenue'])
    / model_assessment['revenue'])

model_assessment['model2_abs_pct_error'] = abs(
    (model_assessment['revenue'] - model_assessment['model2_predicted_revenue'])
    / model_assessment['revenue'])

model1_mape = model_assessment['model1_abs_pct_error'].mean()
model2_mape = model_assessment['model2_abs_pct_error'].mean()

print(f"Model 1 (Winter) MAPE: {model1_mape:.4f}")
print(f"Model 2 (Fall) MAPE: {model2_mape:.4f}")
```

```python
def rmse(actual, predicted):
    return np.sqrt(((actual - predicted) ** 2).mean())

model1_rmse = rmse(model_assessment['revenue'], model_assessment['model1_predicted_revenue'])
model2_rmse = rmse(model_assessment['revenue'], model_assessment['model2_predicted_revenue'])

print(f"Model 1 (Winter) RMSE: {model1_rmse:,.2f}")
print(f"Model 2 (Fall) RMSE: {model2_rmse:,.2f}")
print(f"Model 1 (Winter) R-squared (in-sample, training data): {model1.rsquared:.4f}")
print(f"Model 2 (Fall) R-squared (in-sample, training data): {model2.rsquared:.4f}")
```

```python
plt.figure(figsize=(10,6))
plt.title('Model Comparison: Actual vs Predicted Revenue (Test Data)')
plt.xlabel('Production')
plt.ylabel('Revenue')
plt.scatter(model_assessment['production'], model_assessment['revenue'],
            color='black', label='Actual', s=60, zorder=3)
plt.scatter(model_assessment['production'], model_assessment['model1_predicted_revenue'],
            color='blue', label='Model 1 (Winter) Predicted', marker='x')
plt.scatter(model_assessment['production'], model_assessment['model2_predicted_revenue'],
            color='red', label='Model 2 (Fall) Predicted', marker='x')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.bar(['Model 1 (Winter)', 'Model 2 (Fall)'], [model1_mape, model2_mape], color=['blue','red'])
plt.ylabel('MAPE')
plt.title('MAPE Comparison')
plt.show()
```
